```python
import pandas as pd
RUTA = "https://raw.githubusercontent.com/EmanuelHenaoE/laboratorios-calculo-2026-2/main/datos/sipsa_mensual.csv.gz"
df = pd.read_csv(RUTA)
```

```python
# Filtrar para Manizales, diciembre de 2024 y productos relacionados con plátano
platano_mzl_dic24 = df[
    (df["ciudad"] == "Manizales") &
    (df["fecha"] == "2024-12") &
    (df["producto"].str.contains("plátano", case=False, na=False))
]

print(platano_mzl_dic24[["fecha", "ciudad", "mercado", "producto", "precio_kg"]])
```
